```python
import joblib
```

```python
import pandas as pd
```

```python
loaded_model = joblib.load(
    r"E:\ml assignment\diabetes_dataset_prediction\models\best_model_random_forest.pkl"
)
```

```python
loaded_scaler = joblib.load(
    r"E:\ml assignment\diabetes_dataset_prediction\models/scaler.pkl"
)
```

```python
age = 28

bmi = 22.4

HbA1c_level = 5.2

blood_glucose_level = 92        

hypertension = 0

heart_disease = 0

smoking_history = 1
```

```python
new_patient = pd.DataFrame([{

    'blood_glucose_level': blood_glucose_level,

    'HbA1c_level': HbA1c_level,

    'age': age,

    'bmi': bmi,

    'hypertension': hypertension,

    'heart_disease': heart_disease,

    'smoking_history': smoking_history

}])
```

```python
columns_to_scale = [

    'age',

    'bmi',

    'HbA1c_level',

    'blood_glucose_level'

]

new_patient[columns_to_scale] = loaded_scaler.transform(

    new_patient[columns_to_scale]

)
```

```python
new_patient['risk_score'] = (

    new_patient['hypertension'] +

    new_patient['heart_disease'] +

    (new_patient['smoking_history'] >= 4).astype(int) +

    (new_patient['bmi'] >= 0).astype(int) +

    (new_patient['age'] >= 0).astype(int)

)
```

```python
new_patient['blood_sugar_risk'] = (

    (new_patient['HbA1c_level'] >= 0.9).astype(int) +

    (new_patient['blood_glucose_level'] >= 0.5).astype(int)

)
```

```python
new_patient['bmi_glucose'] = (

    new_patient['bmi'] *

    new_patient['blood_glucose_level']

)
```

```python
new_patient['age_hypertension'] = (

    new_patient['age'] *

    new_patient['hypertension']

)
```

```python
feature_columns = [

    'blood_glucose_level',

    'HbA1c_level',

    'blood_sugar_risk',

    'risk_score',

    'age',

    'bmi_glucose',

    'bmi',

    'hypertension',

    'age_hypertension',

    'heart_disease',

    'smoking_history'

]

new_patient = new_patient[
    feature_columns
]
```

```python
prediction = loaded_model.predict(
    new_patient
)

print(prediction)
```

```python
if prediction[0] == 1:

    print("Patient is Diabetic")

else:

    print("Patient is Non-Diabetic")
```
